```python
#exp 1
import numpy as np

student_scores = np.loadtxt(
    "/content/Students_data.csv",
    delimiter=",",
    skiprows=1,
    usecols=(1, 2, 3, 4)
)

subjects = ["Math", "Science", "English", "History"]

average_scores = np.mean(student_scores, axis=0)

highest_subject = subjects[np.argmax(average_scores)]

print("Average Scores:")
for subject, avg in zip(subjects, average_scores):
    print(f"{subject}: {avg:.2f}")

print("\nHighest Average Subject:", highest_subject)
```

```python
#exp5
import numpy as np

student_data = np.genfromtxt(
    "/content/Students_data.csv",
    delimiter=",",
    skip_header=1,
    dtype=str
)

student_names = student_data[:, 0]
science_scores = student_data[:, 2].astype(float)

# Overall average Science score
overall_average_science_score = np.mean(science_scores)

# Science scores for Alice and Bob
alice_science_score = science_scores[student_names == "Alice"]
bob_science_score = science_scores[student_names == "Bob"]

# Calculate their average Science scores (though for single values, it's just the value)
alice_avg_science = np.mean(alice_science_score)
bob_avg_science = np.mean(bob_science_score)

# Calculate percentage difference (e.g., Alice's score relative to Bob's)
percentage_difference = ((alice_avg_science - bob_avg_science) / bob_avg_science) * 100

print("Overall Average Science Score:", round(overall_average_science_score, 2))
print("Alice's Science Score:", round(alice_avg_science, 2))
print("Bob's Science Score:", round(bob_avg_science, 2))
print("Percentage difference (Alice vs Bob):", round(percentage_difference, 2), "%")
```

```python
import pandas as pd

df = pd.read_csv("/content/Students_data.csv")

# Calculate the total of a subject's score, for example 'Math'
total_math_score = df["Math"].sum()

print("Total Math score for all students:", round(total_math_score, 2))

# Hypothetical bonus and penalty percentages (adapting from discount and tax)
bonus_percentage = 0.10 # e.g., 10% bonus
penalty_percentage = 0.18 # e.g., 18% penalty

# Calculate score after applying a bonus
score_after_bonus = total_math_score + (total_math_score * bonus_percentage)

# Calculate final score after applying a penalty
final_adjusted_score = score_after_bonus - (score_after_bonus * penalty_percentage)

print(
    "Final adjusted Math score (after bonus and penalty):",
    round(final_adjusted_score, 2)
)
```

```python
import pandas as pd

df = pd.read_csv("/content/Students_data.csv")

# Calculate the average score for each student across all subjects
df['Average_Score'] = df[['Math', 'Science', 'English', 'History']].mean(axis=1)

# Group by student and get their average score
student_average_scores = df.groupby("Student")["Average_Score"].mean()

# Find the student with the highest average score
highest_avg_student = student_average_scores.idxmax()
highest_avg_score = student_average_scores.max()

print("Average score for each student:")
print(student_average_scores.to_string())

print(f"\nStudent with the highest average score: {highest_avg_student} (Average Score: {highest_avg_score:.2f})")
```

```python
from collections import Counter
import string

# Create a dummy sample_text.txt file for demonstration
with open("sample_text.txt", "w") as f:
    f.write("This is a sample text. This text will be analyzed for word frequency. sample text.")

with open("sample_text.txt", "r") as file:
    text = file.read()

text = text.lower()
text = text.translate(str.maketrans("", "", string.punctuation))
words = text.split()

frequency = Counter(words)

print("Frequency Distribution of Words:")

for word, count in frequency.items():
    print(word, ":", count)
```

```python
from collections import Counter
import string

reviews = [
    "The product is very good and useful",
    "I really like this product",
    "The product is good and affordable",
    "Very useful product and good quality"
]

words = []

for review in reviews:
    review = review.lower()
    review = review.translate(str.maketrans("", "", string.punctuation))
    words.extend(review.split())

frequency = Counter(words)

print("Frequency Distribution of Words:")

for word, count in frequency.items():
    print(word, ":", count)
```

```python
import pandas as pd
import string
from collections import Counter
import matplotlib.pyplot as plt

# Create a dummy data.csv file with a 'feedback' column
dummy_data = {
    'feedback': [
        'This product is great and very useful.',
        'I love this product, it is amazing.',
        'The product is okay, but could be better.',
        'Not happy with this purchase, it is terrible.',
        'Great product, highly recommend it.'
    ]
}
dummy_df = pd.DataFrame(dummy_data)
dummy_df.to_csv('data.csv', index=False)

df = pd.read_csv('data.csv')

stop_words = {
    "the", "and", "is", "a", "an", "of", "to", "in",
    "for", "on", "with", "this", "that", "it", "was",
    "are", "i", "my", "very", "but", "as", "at", "be"
}

# Process text and filter stop words
words = [word
         for text_entry in df["feedback"].dropna()
         for word in text_entry.lower().translate(str.maketrans("", "", string.punctuation)).split()
         if word not in stop_words]

frequency = Counter(words)

# Set N to a default value for demonstration purposes
n = 5 # You can change this value if you want a different number of top words

top_words = frequency.most_common(n)

print(f"Top {n} Most Frequent Words:")

for word, count in top_words:
    print(f"{word} : {count}")

# Prepare data for plotting
words_list = [word for word, count in top_words]
counts = [count for word, count in top_words]

# Create the bar chart
plt.figure(figsize=(10, 6)) # Increased figure size for better readability
plt.bar(words_list, counts, color='skyblue')
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title(f"Top {n} Most Frequent Words")
plt.xticks(rotation=45, ha='right') # Rotate and align x-axis labels
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats

data = {
    "age": [23, 23, 27, 27, 39, 41, 47, 49, 50, 52, 52, 54, 56, 57, 58, 60, 61, 62],
    "%fat": [9.5, 26.5, 7.8, 17.8, 31.4, 25.9, 27.4, 27.2, 31.2, 34.6, 42.5, 28.8, 33.4, 30.2, 34.1, 32.9, 41.2, 35.7]
}

df = pd.DataFrame(data)

print("Mean:")
print(df.mean())

print("\nMedian:")
print(df.median())

print("\nStandard Deviation:")
print(df.std())

plt.figure(figsize=(8, 5))
plt.boxplot([df["age"], df["%fat"]], labels=["Age", "%Fat"])
plt.title("Boxplot of Age and %Fat")
plt.ylabel("Values")
plt.show()

plt.figure(figsize=(8, 5))
plt.scatter(df["age"], df["%fat"])
plt.xlabel("Age")
plt.ylabel("% Fat")
plt.title("Scatter Plot of Age vs %Fat")
plt.show()

plt.figure(figsize=(8, 5))
stats.probplot(df["age"], dist="norm", plot=plt)
plt.title("Q-Q Plot of Age")
plt.show()

plt.figure(figsize=(8, 5))
stats.probplot(df["%fat"], dist="norm", plot=plt)
plt.title("Q-Q Plot of %Fat")
plt.show()
```

```python
import pandas as pd

# Create a dummy sales_data.csv file
data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Laptop', 'Mouse'],
    'Quantity Sold': [10, 50, 30, 15, 5, 20],
    'Unit Price': [1200, 25, 75, 300, 1250, 28]
}
dummy_sales_df = pd.DataFrame(data)
dummy_sales_df.to_csv('sales_data.csv', index=False)

df = pd.read_csv("sales_data.csv")

df["Total Sales"] = df["Quantity Sold"] * df["Unit Price"]

product_sales = df.groupby("Product")["Total Sales"].sum()

# Assuming a 20% profit margin on total sales
product_profit = product_sales * 0.20

overall_profit = product_profit.sum()

print("Total Sales for Each Product:")
print(product_sales)

print("\nProfit for Each Product:")
print(product_profit)

print(f"\nOverall Profit: {overall_profit:.2f}")

top_5_products = product_sales.sort_values(ascending=False).head(5)

print("\nTop 5 Most Profitable Products (by Total Sales):")
print(top_5_products)
```

```python
import pandas as pd

# Create a dummy customer_data.csv file
dummy_customer_data = {
    'Customer ID': [1, 2, 3, 4, 5, 6],
    'Age': [30, 25, 40, 35, 28, 45],
    'Total Spending': [12000, 4500, 15000, 7000, 3000, 20000]
}
dummy_customer_df = pd.DataFrame(dummy_customer_data)
dummy_customer_df.to_csv('customer_data.csv', index=False)

df = pd.read_csv("customer_data.csv")

def segment(spending):
    if spending >= 10000:
        return "High Spenders"
    elif spending >= 5000:
        return "Medium Spenders"
    else:
        return "Low Spenders"

df["Spending Segment"] = df["Total Spending"].apply(segment)

average_age = df.groupby("Spending Segment")["Age"].mean()

print("Customer Data:")
print(df)

print("\nAverage Age of Customers in Each Spending Segment:")
print(average_age)
```
